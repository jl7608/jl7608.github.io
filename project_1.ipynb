# Life Expectancy Around the World

**Computing in Context Project 1**

Life expectancy is one way to look at changes in public health and living conditions. In this project, I use a dataset of countries over time to answer two simple questions: What does life expectancy look like across the observations, and how has it changed over time?

## 1. Data source

I use the Gapminder dataset, which includes country-level information on life expectancy, population, and GDP per person. It contains 1,704 rows for 142 countries in 12 years from 1952 to 2007.

- **Original data source:** [Gapminder](https://www.gapminder.org/data/)
- **Dataset version used:** [Plotly Express sample data documentation](https://plotly.com/python-api-reference/generated/plotly.express.data.html#plotly.express.data.gapminder)

The CSV file included with this notebook (`gapminder.csv`) comes from the Gapminder dataset packaged with Plotly Express. Each row represents one country in one year. I will focus on **`lifeExp`**, the numeric column for life expectancy at birth in years. I will also use `year` to group the results for the visualization.

## 2. Read the CSV with pandas


```python
import pandas as pd

data = pd.read_csv("gapminder.csv")

print("Number of rows:", len(data))
print("Number of columns:", len(data.columns))
print(data.head())
```

There are several numeric columns in this dataset, but I will use **`lifeExp`** for the three summary statistics. I also check for missing values before doing the calculations.

```python
life_expectancy = data["lifeExp"].dropna()

print("Life expectancy observations:", len(life_expectancy))
print("Missing life expectancy values:", data["lifeExp"].isna().sum())
```

## 3. Mean, median, and mode with pandas


```python
pandas_mean = life_expectancy.mean()
pandas_median = life_expectancy.median()
pandas_modes = sorted(life_expectancy.mode().tolist())

print(f"Mean: {pandas_mean:.3f} years")
print(f"Median: {pandas_median:.3f} years")
print("Mode:", pandas_modes, "years")
```

The mean and median are close. The mode is a specific life expectancy value that occurs more often than any other exact value in the data. Since life expectancy is measured quite precisely, the mode is not as informative here as the mean and median.

## 4. Calculate the same statistics without pandas

```python
import csv

values = []

with open("gapminder.csv", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row["lifeExp"] != "":
            values.append(float(row["lifeExp"]))

print("Values read from CSV:", len(values))
```

### Mean


```python
total = 0

for number in values:
    total += number

manual_mean = total / len(values)
print(f"Mean: {manual_mean:.3f} years")
```

### Median

```python
sorted_values = sorted(values)
n = len(sorted_values)

if n % 2 == 0:
    middle_left = sorted_values[n // 2 - 1]
    middle_right = sorted_values[n // 2]
    manual_median = (middle_left + middle_right) / 2
else:
    manual_median = sorted_values[n // 2]

print(f"Median: {manual_median:.3f} years")
```

### Mode

```python
counts = {}

for number in values:
    if number in counts:
        counts[number] += 1
    else:
        counts[number] = 1

highest_count = max(counts.values())
manual_modes = []

for number in counts:
    if counts[number] == highest_count:
        manual_modes.append(number)

manual_modes.sort()
print("Mode:", manual_modes, "years")
print("Number of times each mode occurs:", highest_count)
```

### Check the results

Both approaches should give the same answers, apart from tiny differences caused by how computers store decimals.

```python
print("Means match:", abs(pandas_mean - manual_mean) < 0.000001)
print("Medians match:", abs(pandas_median - manual_median) < 0.000001)
print("Modes match:", pandas_modes == manual_modes)
```

## 5. Data visualization

I use pandas to calculate average life expectancy by year and basic Python to create a bar chart. Each # represents approximately 2 years.

```python
yearly_average = data.groupby("year")["lifeExp"].mean()

print("Average Life Expectancy by Year")
print("Year | Average life expectancy (years)")
print("     | Each # represents about 2 years")
print("-" * 65)

for year, average in yearly_average.items():
    bar = "#" * round(average / 2)
    print(f"{year} | {bar} {average:.1f}")
```

## 6. What did I find?

The chart shows a clear increase in the average life expectancy of the countries in this dataset between 1952 and 2007. The steady growth in the bars is easier to notice than it would be from a list of numbers alone.

There is an important limitation: **this is an unweighted average of countries**, not the life expectancy of the world's population. Each country has the same influence on the average, regardless of its population. The dataset is also historical, so these results do not describe life expectancy today.
